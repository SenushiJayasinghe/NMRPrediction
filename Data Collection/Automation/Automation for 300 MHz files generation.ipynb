```python
import pyautogui
import pyperclip
import time
import pandas as pd
import os
import shutil
import glob
from datetime import datetime

pyautogui.FAILSAFE = False
pyautogui.PAUSE = 1.0

# ========== CALIBRATION ==========
SMILES_FIELD     = (201, 253)
SIMULATE_BTN     = (268, 692)
JCAMP_BTN        = (1205, 626)

FREQ_DROPDOWN    = (184, 766)
FREQ_300_MHz     = (157, 860)

NPOINTS_DROPDOWN = (182, 863)
NPOINTS_4K       = (158, 674)

CLEAR_STRUCTURE  = (15, 344)
REFRESH_BTN      = (96, 62)

SAVE_FOLDER      = r"C:\Users\pljh0187\Desktop\Senushi\JCAMP Files for 300MHz"
DOWNLOADS        = r"C:\Users\pljh0187\Downloads"
LOG_FILE         = os.path.join(r"C:\Users\pljh0187\Desktop\Senushi", "automation_log_300MHz.txt")

# ========== READ Book4.csv ==========
CSV_PATH = r"C:\Users\pljh0187\Desktop\Senushi\Book4.csv"
df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

if 'ID' not in df.columns:
    raise KeyError("Book4.csv must have an 'ID' column")
if 'Name' not in df.columns:
    raise KeyError("Book4.csv must have a 'Name' column")
if 'SMILES' not in df.columns:
    raise KeyError("Book4.csv must have a 'SMILES' column")

df['ID'] = df['ID'].astype(str).str.zfill(3)

compounds = list(zip(df['ID'], df['Name'].str.strip(), df['SMILES'].str.strip()))
total = len(compounds)
print(f"Loaded {total} compounds from {CSV_PATH}")

os.makedirs(SAVE_FOLDER, exist_ok=True)

# ==========================================================
# FILE NAMING LOGIC:
# Each downloaded JCAMP file is renamed to:
#     {compound_id}_300MHz.jdx
# where compound_id is the zero-padded 3-digit ID from Book4.
# ==========================================================

def safe_filename(compound_id):
    return f"{compound_id}_300MHz.jdx"

def log(message):
    ts = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    line = f"[{ts}] {message}"
    print(line)
    with open(LOG_FILE, "a", encoding="utf-8") as f:
        f.write(line + "\n")

def wait_for_download(before_files, timeout=2):
    for _ in range(timeout * 2):
        time.sleep(0.5)
        current = set(glob.glob(os.path.join(DOWNLOADS, "*.jdx")) +
                      glob.glob(os.path.join(DOWNLOADS, "predictedSpectrum*")))
        new_files = current - before_files
        if new_files:
            return new_files.pop()
    return None

def rename_download(downloaded_path, compound_id):
    filename = safe_filename(compound_id)
    dest = os.path.join(SAVE_FOLDER, filename)
    shutil.move(downloaded_path, dest)
    return dest

def verify_downloaded_freq(path, expected_freq=300):
    with open(path, 'r', encoding='utf-8', errors='replace') as f:
        for line in f:
            line = line.strip()
            if line.startswith('##.OBSERVE FREQUENCY='):
                actual = line.split('=')[1].strip()
                return actual, str(expected_freq) == actual
            if line == '##END=':
                break
    return 'unknown', False

def hover_and_paste(x, y, text):
    pyperclip.copy(text)
    pyautogui.moveTo(x, y, duration=0.6)
    time.sleep(0.4)
    pyautogui.hotkey('ctrl', 'v')
    time.sleep(1.0)

def hover_and_click(x, y):
    pyautogui.moveTo(x, y, duration=0.6)
    pyautogui.click()

def clear_smiles_field():
    pyautogui.moveTo(*SMILES_FIELD, duration=0.6)
    pyautogui.hotkey('ctrl', 'a')
    pyautogui.press('delete')

def select_300MHz():
    hover_and_click(*FREQ_DROPDOWN)
    hover_and_click(*FREQ_300_MHz)

def select_4k_points():
    hover_and_click(*NPOINTS_DROPDOWN)
    hover_and_click(*NPOINTS_4K)

def simulate_and_download(compound_id, name, smiles):
    before = set(glob.glob(os.path.join(DOWNLOADS, "*.jdx")) +
                 glob.glob(os.path.join(DOWNLOADS, "predictedSpectrum*")))

    # Step 1 — clear SMILES field and paste new SMILES
    clear_smiles_field()
    hover_and_paste(*SMILES_FIELD, smiles)

    # Step 2 — select 300 MHz
    select_300MHz()

    # Step 3 — select 4k points
    select_4k_points()

    # Step 4 — click Simulate, then wait 4 seconds
    hover_and_click(*SIMULATE_BTN)
    time.sleep(4)

    # Step 5 — download JCAMP
    hover_and_click(*JCAMP_BTN)

    # Step 6 — wait for file to appear in Downloads
    downloaded = wait_for_download(before, timeout=2)

    if downloaded:
        actual_freq, freq_ok = verify_downloaded_freq(downloaded, expected_freq=300)
        dest = rename_download(downloaded, compound_id)

        if not freq_ok:
            log(f"WARNING | ID={compound_id} | {name} | FREQUENCY MISMATCH "
                f"— file says {actual_freq} MHz | saved as: {os.path.basename(dest)}")
        else:
            log(f"SUCCESS | ID={compound_id} | {name} | 300 MHz | "
                f"saved as: {os.path.basename(dest)}")
    else:
        log(f"WARNING | ID={compound_id} | {name} | 300 MHz | "
            f"file not found in Downloads after 2s")

    # Step 7 — clear structure and refresh page
    hover_and_click(*CLEAR_STRUCTURE)
    hover_and_click(*REFRESH_BTN)

    # Step 8 — wait 1 minute for page to reset
    log(f"Waiting 1 minute for page to reset...")
    time.sleep(60)

    # *** Step 9 — click X again after 1 minute rest to clear structure ***
    log(f"Clicking clear structure again after rest...")
    hover_and_click(*CLEAR_STRUCTURE)

# ========== MAIN RUN ==========
log("=" * 60)
log(f"Starting 300 MHz NMR automation — {total} compounds")
log(f"Save folder : {SAVE_FOLDER}")
log(f"CSV source  : {CSV_PATH}")
log("=" * 60)

print(f"Loaded {total} compounds. Files will be saved to: {SAVE_FOLDER}")
for i, (cid, name, smi) in enumerate(compounds, start=1):
    print(f"  [{i:03d}] ID={cid} : {name}")

print("\nMake sure the cheminfo.org page is open and visible.")
print("You have 5 seconds to click on the browser window...")
for i in range(5, 0, -1):
    print(f"  Starting in {i}...")
    time.sleep(1)
print("Starting now!")

start_time    = time.time()
success_count = 0
warning_count = 0
error_count   = 0

for idx, (cid, name, smiles) in enumerate(compounds, start=1):
    print(f"\n{'='*60}")
    log(f"Compound {idx}/{total} : ID={cid} | {name}")
    log(f"SMILES : {smiles}")

    if idx > 1:
        elapsed   = time.time() - start_time
        avg_per   = elapsed / (idx - 1)
        remaining = avg_per * (total - idx + 1) / 60
        log(f"Estimated time remaining: ~{remaining:.0f} minutes")

    try:
        simulate_and_download(cid, name, smiles)
        success_count += 1
        log(f"Finished: ID={cid} — {name}")

    except Exception as e:
        log(f"ERROR | ID={cid} | {name} | {e} — skipping to next compound")
        error_count += 1
        continue

elapsed_total = (time.time() - start_time) / 60
log("=" * 60)
log(f"All done!")
log(f"  Compounds processed : {total}")
log(f"  Total files         : {success_count}")
log(f"  Warnings            : {warning_count}")
log(f"  Errors              : {error_count}")
log(f"  Save location       : {SAVE_FOLDER}")
log(f"  Total time          : {elapsed_total:.1f} minutes")
log(f"  Log saved to        : {LOG_FILE}")
log("=" * 60)
```

```python
import nmrglue as ng
import numpy as np
import matplotlib.pyplot as plt
import glob
import os

# ========== FOLDER ==========
folder = r"C:\Users\pljh0187\Desktop\Senushi\JCAMP Files for 300MHz"
jdx_files = sorted(glob.glob(os.path.join(folder, "*.jdx")))

print(f"Found {len(jdx_files)} files:")
for f in jdx_files:
    print(f"  {os.path.basename(f)}")

# ========== HELPERS ==========
def safe_float(val, default=0.0):
    if isinstance(val, list):
        val = val[0] if val else default
    try:
        return float(val)
    except:
        return default

def read_jdx(path):
    # Read YFACTOR and frequency directly from raw file
    yfactor  = 1.0
    obs_freq = 'unknown'
    npoints_header = None

    with open(path, 'r', encoding='utf-8', errors='replace') as f:
        for line in f:
            line = line.strip()
            if line.startswith('##.OBSERVE FREQUENCY='):
                obs_freq = line.split('=')[1].strip()
            elif line.startswith('##YFACTOR='):
                try:
                    yfactor = float(line.split('=')[1].strip())
                except:
                    pass
            elif line.startswith('##NPOINTS='):
                try:
                    npoints_header = int(line.split('=')[1].strip())
                except:
                    pass
            if line == '##END=':
                break

    # Read with nmrglue
    dic, data = ng.jcampdx.read(path)
    data = np.asarray(data, dtype=np.float64)

    # Apply YFACTOR only if nmrglue hasn't (raw integers > 1e8)
    if np.abs(data).max() > 1e8:
        data = data * yfactor

    # Build x axis
    npoints = len(data)
    firstx  = safe_float(dic.get('FIRSTX'), 0.0)
    lastx   = safe_float(dic.get('LASTX'),  0.0)
    deltax  = safe_float(dic.get('DELTAX'), 0.0)

    if deltax == 0.0 and npoints > 1:
        deltax = (lastx - firstx) / (npoints - 1)

    x = firstx + np.arange(npoints, dtype=np.float64) * deltax

    # Reverse to descending ppm (NMR convention)
    if x[0] < x[-1]:
        x    = x[::-1].copy()
        data = data[::-1].copy()

    return x, data, obs_freq, npoints, npoints_header

# ========== PLOT ==========
n_files = len(jdx_files)
fig, axes = plt.subplots(n_files, 1, figsize=(14, 4 * n_files))

# Handle case where only 1 file
if n_files == 1:
    axes = [axes]

fig.suptitle('Predicted 300 MHz NMR Spectra', fontsize=15, fontweight='bold', y=1.01)

for ax, path in zip(axes, jdx_files):
    fname = os.path.basename(path)
    compound_id = fname.replace('_300MHz.jdx', '')

    try:
        x, data, obs_freq, npoints_decoded, npoints_header = read_jdx(path)

        # Check all 4096 points decoded
        points_ok = (npoints_decoded == 4096)
        points_note = f"{npoints_decoded} pts decoded"
        if not points_ok:
            points_note += f" ⚠ (expected {npoints_header})"

        ax.plot(x, data, color='black', linewidth=0.7)
        ax.set_title(
            f"Compound {compound_id} — {obs_freq} MHz | {points_note}",
            fontsize=10, fontweight='bold'
        )
        ax.set_xlabel('Chemical Shift (ppm)', fontsize=9)
        ax.set_ylabel('Intensity', fontsize=9)
        ax.invert_xaxis()
        ax.set_xlim(max(x), min(x))
        ax.grid(True, alpha=0.3)
        ax.tick_params(labelsize=8)

        print(f"  {fname} | freq={obs_freq} MHz | points={npoints_decoded} "
              f"| x range={x.min():.2f}–{x.max():.2f} ppm "
              f"| max intensity={data.max():.3e}")

    except Exception as e:
        ax.text(0.5, 0.5, f"Error reading {fname}:\n{e}",
                ha='center', va='center', transform=ax.transAxes,
                color='red', fontsize=9)
        ax.set_title(fname)
        print(f"  ERROR: {fname} — {e}")

plt.tight_layout()

out_path = r"C:\Users\pljh0187\Desktop\Senushi\300MHz_spectra.png"
plt.savefig(out_path, dpi=150, bbox_inches='tight')
plt.show()
print(f"\nPlot saved to: {out_path}")
```

```python
import os
import glob
import numpy as np
import nmrglue as ng

# ================= FOLDER =================
folder = r"C:\Users\pljh0187\Desktop\Senushi\JCAMP Files for 300MHz"

# Expected file count (based on naming convention)
expected_total = 502

# Actual files in folder
jdx_files = sorted(glob.glob(os.path.join(folder, "*.jdx")))

print("=" * 60)
print(f"Files found in folder: {len(jdx_files)}")
print(f"Expected files       : {expected_total}")
print("=" * 60)

# ================= TRACKERS =================
missing_files = []
null_text_files = []
empty_spectrum_files = []
read_error_files = []

# Build expected filenames
expected_files = [
    os.path.join(folder, f"{i:03d}_300MHz.jdx")
    for i in range(1, expected_total + 1)
]

actual_set = set(jdx_files)

# ================= 1. CHECK MISSING FILES =================
for f in expected_files:
    if f not in actual_set:
        missing_files.append(os.path.basename(f))

# ================= 2. CHECK CONTENT =================
for path in jdx_files:
    fname = os.path.basename(path)

    try:
        # -------- check raw text for "null" --------
        with open(path, "r", encoding="utf-8", errors="ignore") as f:
            text = f.read().lower()

        if "null" in text:
            null_text_files.append(fname)
            continue

        # -------- try reading spectrum --------
        dic, data = ng.jcampdx.read(path)
        data = np.asarray(data, dtype=np.float64)

        # empty spectrum
        if data.size == 0:
            empty_spectrum_files.append(fname)
            continue

        # all zeros or invalid
        if np.all(np.isclose(data, 0)) or np.all(np.isnan(data)):
            empty_spectrum_files.append(fname)
            continue

    except Exception as e:
        read_error_files.append((fname, str(e)))

# ================= SUMMARY =================
print("\n" + "=" * 60)
print("SUMMARY")
print("=" * 60)

print(f"Missing files           : {len(missing_files)}")
print(f'Files containing "null" : {len(null_text_files)}')
print(f"Empty/invalid spectra   : {len(empty_spectrum_files)}")
print(f"Read errors             : {len(read_error_files)}")

print("=" * 60)

# ================= DETAILS =================
if missing_files:
    print("\n MISSING FILES:")
    for f in missing_files:
        print(f)

if null_text_files:
    print("\n FILES CONTAINING 'null':")
    for f in null_text_files:
        print(f)

if empty_spectrum_files:
    print("\n EMPTY / INVALID SPECTRA:")
    for f in empty_spectrum_files:
        print(f)

if read_error_files:
    print("\n READ ERRORS:")
    for f, err in read_error_files:
        print(f"{f} -> {err}")

if not (missing_files or null_text_files or empty_spectrum_files or read_error_files):
    print("\n All files are valid and complete.")
```

```python
import pyautogui
import pyperclip
import time
import pandas as pd
import os
import shutil
import glob
from datetime import datetime

pyautogui.FAILSAFE = False
pyautogui.PAUSE = 1.0

# ========== CALIBRATION ==========
SMILES_FIELD     = (201, 253)
SIMULATE_BTN     = (268, 692)
JCAMP_BTN        = (1205, 626)

FREQ_DROPDOWN    = (184, 766)
FREQ_300_MHz     = (157, 860)

NPOINTS_DROPDOWN = (182, 863)
NPOINTS_4K       = (158, 674)

CLEAR_STRUCTURE  = (15, 344)
REFRESH_BTN      = (96, 62)

SAVE_FOLDER      = r"C:\Users\pljh0187\Desktop\Senushi\JCAMP Files for 300MHz"
DOWNLOADS        = r"C:\Users\pljh0187\Downloads"
LOG_FILE         = os.path.join(r"C:\Users\pljh0187\Desktop\Senushi", "automation_log_300MHz.txt")

# ========== READ Book4.csv ==========
CSV_PATH = r"C:\Users\pljh0187\Desktop\Senushi\Book4.csv"
df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

if 'ID' not in df.columns:
    raise KeyError("Book4.csv must have an 'ID' column")
if 'Name' not in df.columns:
    raise KeyError("Book4.csv must have a 'Name' column")
if 'SMILES' not in df.columns:
    raise KeyError("Book4.csv must have a 'SMILES' column")

df['ID'] = df['ID'].astype(str).str.zfill(3)

all_compounds = list(zip(df['ID'], df['Name'].str.strip(), df['SMILES'].str.strip()))
print(f"Loaded {len(all_compounds)} compounds total from {CSV_PATH}")

os.makedirs(SAVE_FOLDER, exist_ok=True)

# ==========================================================
# FILE NAMING LOGIC:
# Each downloaded JCAMP file is renamed to:
#     {compound_id}_300MHz.jdx
# where compound_id is the zero-padded 3-digit ID from Book4.
# ==========================================================

def safe_filename(compound_id):
    return f"{compound_id}_300MHz.jdx"

# ========== CHECK WHICH FILES ALREADY EXIST ==========
existing_files = set(os.path.basename(f) for f in glob.glob(os.path.join(SAVE_FOLDER, "*.jdx")))
print(f"Found {len(existing_files)} existing files in: {SAVE_FOLDER}")

missing_compounds = []
for cid, name, smiles in all_compounds:
    expected_fname = safe_filename(cid)
    if expected_fname not in existing_files:
        missing_compounds.append((cid, name, smiles))

print(f"Missing compounds : {len(missing_compounds)}")

# *** Only take the FIRST 30 missing compounds ***
compounds = missing_compounds[:30]
total = len(compounds)

print(f"\nWill download {total} missing files (first 30 missing):")
for i, (cid, name, smi) in enumerate(compounds, start=1):
    print(f"  [{i:03d}] ID={cid} : {name}")

# ===========================================================

def log(message):
    ts = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    line = f"[{ts}] {message}"
    print(line)
    with open(LOG_FILE, "a", encoding="utf-8") as f:
        f.write(line + "\n")

def wait_for_download(before_files, timeout=2):
    for _ in range(timeout * 2):
        time.sleep(0.5)
        current = set(glob.glob(os.path.join(DOWNLOADS, "*.jdx")) +
                      glob.glob(os.path.join(DOWNLOADS, "predictedSpectrum*")))
        new_files = current - before_files
        if new_files:
            return new_files.pop()
    return None

def rename_download(downloaded_path, compound_id):
    filename = safe_filename(compound_id)
    dest = os.path.join(SAVE_FOLDER, filename)
    shutil.move(downloaded_path, dest)
    return dest

def verify_downloaded_freq(path, expected_freq=300):
    with open(path, 'r', encoding='utf-8', errors='replace') as f:
        for line in f:
            line = line.strip()
            if line.startswith('##.OBSERVE FREQUENCY='):
                actual = line.split('=')[1].strip()
                return actual, str(expected_freq) == actual
            if line == '##END=':
                break
    return 'unknown', False

def hover_and_paste(x, y, text):
    pyperclip.copy(text)
    pyautogui.moveTo(x, y, duration=0.6)
    time.sleep(0.4)
    pyautogui.hotkey('ctrl', 'v')
    time.sleep(1.0)

def hover_and_click(x, y):
    pyautogui.moveTo(x, y, duration=0.6)
    pyautogui.click()

def clear_smiles_field():
    pyautogui.moveTo(*SMILES_FIELD, duration=0.6)
    pyautogui.hotkey('ctrl', 'a')
    pyautogui.press('delete')

def select_300MHz():
    hover_and_click(*FREQ_DROPDOWN)
    hover_and_click(*FREQ_300_MHz)

def select_4k_points():
    hover_and_click(*NPOINTS_DROPDOWN)
    hover_and_click(*NPOINTS_4K)

def simulate_and_download(compound_id, name, smiles):
    before = set(glob.glob(os.path.join(DOWNLOADS, "*.jdx")) +
                 glob.glob(os.path.join(DOWNLOADS, "predictedSpectrum*")))

    # Step 1 — clear SMILES field and paste new SMILES
    clear_smiles_field()
    hover_and_paste(*SMILES_FIELD, smiles)

    # Step 2 — select 300 MHz
    select_300MHz()

    # Step 3 — select 4k points
    select_4k_points()

    # Step 4 — click Simulate, then wait 4 seconds
    hover_and_click(*SIMULATE_BTN)
    time.sleep(4)

    # Step 5 — download JCAMP
    hover_and_click(*JCAMP_BTN)

    # Step 6 — wait for file to appear in Downloads
    downloaded = wait_for_download(before, timeout=2)

    if downloaded:
        actual_freq, freq_ok = verify_downloaded_freq(downloaded, expected_freq=300)
        dest = rename_download(downloaded, compound_id)

        if not freq_ok:
            log(f"WARNING | ID={compound_id} | {name} | FREQUENCY MISMATCH "
                f"— file says {actual_freq} MHz | saved as: {os.path.basename(dest)}")
        else:
            log(f"SUCCESS | ID={compound_id} | {name} | 300 MHz | "
                f"saved as: {os.path.basename(dest)}")
    else:
        log(f"WARNING | ID={compound_id} | {name} | 300 MHz | "
            f"file not found in Downloads after 2s")

    # Step 7 — clear structure and refresh page
    hover_and_click(*CLEAR_STRUCTURE)
    hover_and_click(*REFRESH_BTN)

    # Step 8 — wait 30 seconds for page to reset
    log(f"Waiting 30 seconds for page to reset...")
    time.sleep(30)

    # Step 9 — click X again after 30 seconds rest to clear structure
    log(f"Clicking clear structure again after rest...")
    hover_and_click(*CLEAR_STRUCTURE)

# ========== MAIN RUN ==========
log("=" * 60)
log(f"Resuming 300 MHz NMR automation — downloading {total} missing compounds (first 30 missing)")
log(f"Save folder : {SAVE_FOLDER}")
log(f"CSV source  : {CSV_PATH}")
log(f"Existing files found : {len(existing_files)}")
log(f"Total missing files  : {len(missing_compounds)}")
log("=" * 60)

if total == 0:
    log("No missing compounds to download. Exiting.")
else:
    print("\nMake sure the cheminfo.org page is open and visible.")
    print("You have 5 seconds to click on the browser window...")
    for i in range(5, 0, -1):
        print(f"  Starting in {i}...")
        time.sleep(1)
    print("Starting now!")

    start_time    = time.time()
    success_count = 0
    warning_count = 0
    error_count   = 0

    for idx, (cid, name, smiles) in enumerate(compounds, start=1):
        print(f"\n{'='*60}")
        log(f"Compound {idx}/{total} : ID={cid} | {name}")
        log(f"SMILES : {smiles}")

        if idx > 1:
            elapsed   = time.time() - start_time
            avg_per   = elapsed / (idx - 1)
            remaining = avg_per * (total - idx + 1) / 60
            log(f"Estimated time remaining: ~{remaining:.0f} minutes")

        try:
            simulate_and_download(cid, name, smiles)
            success_count += 1
            log(f"Finished: ID={cid} — {name}")

        except Exception as e:
            log(f"ERROR | ID={cid} | {name} | {e} — skipping to next compound")
            error_count += 1
            continue

    elapsed_total = (time.time() - start_time) / 60
    log("=" * 60)
    log(f"All done!")
    log(f"  Compounds processed : {total}")
    log(f"  Total files         : {success_count}")
    log(f"  Warnings            : {warning_count}")
    log(f"  Errors              : {error_count}")
    log(f"  Save location       : {SAVE_FOLDER}")
    log(f"  Total time          : {elapsed_total:.1f} minutes")
    log(f"  Log saved to        : {LOG_FILE}")
    log("=" * 60)
```

```python
import pyautogui
import pyperclip
import time
import pandas as pd
import os
import shutil
import glob
from datetime import datetime

pyautogui.FAILSAFE = False
pyautogui.PAUSE = 1.0

# ========== CALIBRATION ==========
SMILES_FIELD     = (201, 253)
SIMULATE_BTN     = (268, 692)
JCAMP_BTN        = (1205, 626)

FREQ_DROPDOWN    = (184, 766)
FREQ_300_MHz     = (157, 860)

NPOINTS_DROPDOWN = (182, 863)
NPOINTS_4K       = (158, 674)

CLEAR_STRUCTURE  = (15, 344)
REFRESH_BTN      = (96, 62)

SAVE_FOLDER      = r"C:\Users\pljh0187\Desktop\Senushi\JCAMP Files for 300MHz"
DOWNLOADS        = r"C:\Users\pljh0187\Downloads"
LOG_FILE         = os.path.join(r"C:\Users\pljh0187\Desktop\Senushi", "automation_log_300MHz.txt")

# ========== READ Book4.csv ==========
CSV_PATH = r"C:\Users\pljh0187\Desktop\Senushi\Book4.csv"
df = pd.read_csv(CSV_PATH)
df.columns = df.columns.str.strip()

if 'ID' not in df.columns:
    raise KeyError("Book4.csv must have an 'ID' column")
if 'Name' not in df.columns:
    raise KeyError("Book4.csv must have a 'Name' column")
if 'SMILES' not in df.columns:
    raise KeyError("Book4.csv must have a 'SMILES' column")

df['ID'] = df['ID'].astype(str).str.zfill(3)

all_compounds = list(zip(df['ID'], df['Name'].str.strip(), df['SMILES'].str.strip()))
print(f"Loaded {len(all_compounds)} compounds total from {CSV_PATH}")

os.makedirs(SAVE_FOLDER, exist_ok=True)

# ==========================================================
# FILE NAMING LOGIC:
# Each downloaded JCAMP file is renamed to:
#     {compound_id}_300MHz.jdx
# where compound_id is the zero-padded 3-digit ID from Book4.
# ==========================================================

def safe_filename(compound_id):
    return f"{compound_id}_300MHz.jdx"

# ========== CHECK WHICH FILES ALREADY EXIST ==========
existing_files = set(os.path.basename(f) for f in glob.glob(os.path.join(SAVE_FOLDER, "*.jdx")))
print(f"Found {len(existing_files)} existing files in: {SAVE_FOLDER}")

missing_compounds = []
for cid, name, smiles in all_compounds:
    expected_fname = safe_filename(cid)
    if expected_fname not in existing_files:
        missing_compounds.append((cid, name, smiles))

print(f"Missing compounds : {len(missing_compounds)}")

# *** Only take the NEXT 20 missing compounds ***
compounds = missing_compounds[:140]
total = len(compounds)

print(f"\nWill download {total} missing files (next 20 missing):")
for i, (cid, name, smi) in enumerate(compounds, start=1):
    print(f"  [{i:03d}] ID={cid} : {name}")

# ===========================================================

def log(message):
    ts = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    line = f"[{ts}] {message}"
    print(line)
    with open(LOG_FILE, "a", encoding="utf-8") as f:
        f.write(line + "\n")

def wait_for_download(before_files, timeout=2):
    for _ in range(timeout * 2):
        time.sleep(0.5)
        current = set(glob.glob(os.path.join(DOWNLOADS, "*.jdx")) +
                      glob.glob(os.path.join(DOWNLOADS, "predictedSpectrum*")))
        new_files = current - before_files
        if new_files:
            return new_files.pop()
    return None

def rename_download(downloaded_path, compound_id):
    filename = safe_filename(compound_id)
    dest = os.path.join(SAVE_FOLDER, filename)
    shutil.move(downloaded_path, dest)
    return dest

def verify_downloaded_freq(path, expected_freq=300):
    with open(path, 'r', encoding='utf-8', errors='replace') as f:
        for line in f:
            line = line.strip()
            if line.startswith('##.OBSERVE FREQUENCY='):
                actual = line.split('=')[1].strip()
                return actual, str(expected_freq) == actual
            if line == '##END=':
                break
    return 'unknown', False

def hover_and_paste(x, y, text):
    pyperclip.copy(text)
    pyautogui.moveTo(x, y, duration=0.6)
    time.sleep(0.4)
    pyautogui.hotkey('ctrl', 'v')
    time.sleep(1.0)

def hover_and_click(x, y):
    pyautogui.moveTo(x, y, duration=0.6)
    pyautogui.click()

def clear_smiles_field():
    pyautogui.moveTo(*SMILES_FIELD, duration=0.6)
    pyautogui.hotkey('ctrl', 'a')
    pyautogui.press('delete')

def select_300MHz():
    hover_and_click(*FREQ_DROPDOWN)
    hover_and_click(*FREQ_300_MHz)

def select_4k_points():
    hover_and_click(*NPOINTS_DROPDOWN)
    hover_and_click(*NPOINTS_4K)

def simulate_and_download(compound_id, name, smiles):
    before = set(glob.glob(os.path.join(DOWNLOADS, "*.jdx")) +
                 glob.glob(os.path.join(DOWNLOADS, "predictedSpectrum*")))

    # Step 1 — clear SMILES field and paste new SMILES
    clear_smiles_field()
    hover_and_paste(*SMILES_FIELD, smiles)

    # Step 2 — select 300 MHz
    select_300MHz()

    # Step 3 — select 4k points
    select_4k_points()

    # Step 4 — click Simulate, then wait 4 seconds
    hover_and_click(*SIMULATE_BTN)
    time.sleep(4)

    # Step 5 — download JCAMP
    hover_and_click(*JCAMP_BTN)

    # Step 6 — wait for file to appear in Downloads
    downloaded = wait_for_download(before, timeout=2)

    if downloaded:
        actual_freq, freq_ok = verify_downloaded_freq(downloaded, expected_freq=300)
        dest = rename_download(downloaded, compound_id)

        if not freq_ok:
            log(f"WARNING | ID={compound_id} | {name} | FREQUENCY MISMATCH "
                f"— file says {actual_freq} MHz | saved as: {os.path.basename(dest)}")
        else:
            log(f"SUCCESS | ID={compound_id} | {name} | 300 MHz | "
                f"saved as: {os.path.basename(dest)}")
    else:
        log(f"WARNING | ID={compound_id} | {name} | 300 MHz | "
            f"file not found in Downloads after 2s")

    # Step 7 — clear structure and refresh page
    hover_and_click(*CLEAR_STRUCTURE)
    hover_and_click(*REFRESH_BTN)

    # Step 8 — wait 5 seconds for page to reset
    log(f"Waiting 5 seconds for page to reset...")
    time.sleep(2)

    # Step 9 — click X again after 5 seconds rest to clear structure
    log(f"Clicking clear structure again after rest...")
    hover_and_click(*CLEAR_STRUCTURE)

# ========== MAIN RUN ==========
log("=" * 60)
log(f"Resuming 300 MHz NMR automation — downloading {total} missing compounds (next 20 missing)")
log(f"Save folder : {SAVE_FOLDER}")
log(f"CSV source  : {CSV_PATH}")
log(f"Existing files found : {len(existing_files)}")
log(f"Total missing files  : {len(missing_compounds)}")
log("=" * 60)

if total == 0:
    log("No missing compounds to download. Exiting.")
else:
    print("\nMake sure the cheminfo.org page is open and visible.")
    print("You have 5 seconds to click on the browser window...")
    for i in range(5, 0, -1):
        print(f"  Starting in {i}...")
        time.sleep(1)
    print("Starting now!")

    start_time    = time.time()
    success_count = 0
    warning_count = 0
    error_count   = 0

    for idx, (cid, name, smiles) in enumerate(compounds, start=1):
        print(f"\n{'='*60}")
        log(f"Compound {idx}/{total} : ID={cid} | {name}")
        log(f"SMILES : {smiles}")

        if idx > 1:
            elapsed   = time.time() - start_time
            avg_per   = elapsed / (idx - 1)
            remaining = avg_per * (total - idx + 1) / 60
            log(f"Estimated time remaining: ~{remaining:.0f} minutes")

        try:
            simulate_and_download(cid, name, smiles)
            success_count += 1
            log(f"Finished: ID={cid} — {name}")

        except Exception as e:
            log(f"ERROR | ID={cid} | {name} | {e} — skipping to next compound")
            error_count += 1
            continue

    elapsed_total = (time.time() - start_time) / 60
    log("=" * 60)
    log(f"All done!")
    log(f"  Compounds processed : {total}")
    log(f"  Total files         : {success_count}")
    log(f"  Warnings            : {warning_count}")
    log(f"  Errors              : {error_count}")
    log(f"  Save location       : {SAVE_FOLDER}")
    log(f"  Total time          : {elapsed_total:.1f} minutes")
    log(f"  Log saved to        : {LOG_FILE}")
    log("=" * 60)
```
