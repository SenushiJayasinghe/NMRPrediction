```python
import os
import glob
import numpy as np
import pandas as pd
from pathlib import Path
from datetime import datetime

# ========== CONFIGURATION ==========
SOURCE_FOLDER = r"C:\Users\pljh0187\Desktop\Senushi"
OUTPUT_FOLDER = r"C:\Users\pljh0187\Desktop\Senushi\CSV"
LOG_FILE      = os.path.join(SOURCE_FOLDER, "conversion_log.txt")
# ====================================

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

def log(message):
    ts = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    line = f"[{ts}] {message}"
    print(line)
    with open(LOG_FILE, "a", encoding="utf-8") as f:
        f.write(line + "\n")

# ---------------------------------------------------------------------------
# JCAMP-DX character encoding tables (JCAMP-DX 4.24 standard)
#   SQZ : encode absolute Y value   @ A-I a-i  →  0, +1..+9, -1..-9
#   DIF : encode difference from previous Y   % J-R j-r  →  0, +1..+9, -1..-9
#   DUP : repeat last value N times   S-Z s  →  1..9
# ---------------------------------------------------------------------------
SQZ = {
    '@': 0,
    'A': 1, 'B': 2, 'C': 3, 'D': 4, 'E': 5,
    'F': 6, 'G': 7, 'H': 8, 'I': 9,
    'a':-1, 'b':-2, 'c':-3, 'd':-4, 'e':-5,
    'f':-6, 'g':-7, 'h':-8, 'i':-9,
}
DIF = {
    '%': 0,
    'J': 1, 'K': 2, 'L': 3, 'M': 4, 'N': 5,
    'O': 6, 'P': 7, 'Q': 8, 'R': 9,
    'j':-1, 'k':-2, 'l':-3, 'm':-4, 'n':-5,
    'o':-6, 'p':-7, 'q':-8, 'r':-9,
}
DUP = {'S':1,'T':2,'U':3,'V':4,'W':5,'X':6,'Y':7,'Z':8,'s':9}


def tokenize(line):
    """
    Tokenize one JCAMP data line into (type, value) pairs.
    Types: 'X' (first token = X position), 'SQZ' (absolute Y),
           'DIF' (Y difference), 'DUP' (repeat count).
    """
    tokens = []
    i = 0
    first_token = True

    while i < len(line):
        c = line[i]

        if c in (' ', '\t', ','):
            i += 1
            continue

        if c in SQZ:
            sign   = -1 if (c.islower() and c != '@') else 1
            digits = str(abs(SQZ[c]))
            i += 1
            while i < len(line) and (line[i].isdigit() or line[i] == '.'):
                digits += line[i]; i += 1
            val = sign * float(digits)
            tokens.append(('X' if first_token else 'SQZ', val))
            first_token = False

        elif c in DIF:
            sign   = -1 if c.islower() else 1
            digits = str(abs(DIF[c]))
            i += 1
            while i < len(line) and (line[i].isdigit() or line[i] == '.'):
                digits += line[i]; i += 1
            val = sign * float(digits)
            tokens.append(('DIF', val))
            first_token = False

        elif c in DUP:
            digits = str(DUP[c])
            i += 1
            while i < len(line) and line[i].isdigit():
                digits += line[i]; i += 1
            tokens.append(('DUP', int(digits)))

        elif c.isdigit() or c in ('+', '-'):
            digits = c; i += 1
            while i < len(line) and (line[i].isdigit() or line[i] == '.'):
                digits += line[i]; i += 1
            val = float(digits)
            tokens.append(('X' if first_token else 'SQZ', val))
            first_token = False

        else:
            i += 1  # skip unknown characters

    return tokens


def parse_jcamp(filepath):
    """
    Parse a JCAMP-DX NMR file.
    Returns a DataFrame with columns: chemical_shift_ppm, intensity
    Handles:
      - XYDATA=(X++(Y..Y))  with SQZ / DIF / DUP compression  ← cheminfo.org format
      - XYPOINTS=(XY..XY)   simple X,Y pairs
    The X values in the data block are raw integer indices; the real ppm value
    is  index × deltax  (deltax is already in ppm units in these files).
    """
    with open(filepath, "r", encoding="utf-8", errors="ignore") as f:
        raw_lines = f.readlines()

    # --- parse header ---
    xfactor  = 1.0
    yfactor  = 1.0
    npoints  = None
    deltax   = None
    firstx   = None
    lastx    = None
    data_fmt = None
    data_lines   = []
    xypts_lines  = []
    in_xydata    = False
    in_xypoints  = False

    for line in raw_lines:
        s = line.rstrip("\n").strip()
        u = s.upper()

        if   u.startswith("##XFACTOR="):  xfactor = float(s.split("=",1)[1])
        elif u.startswith("##YFACTOR="):  yfactor = float(s.split("=",1)[1])
        elif u.startswith("##NPOINTS="):  npoints = int(s.split("=",1)[1])
        elif u.startswith("##DELTAX="):   deltax  = float(s.split("=",1)[1])
        elif u.startswith("##FIRSTX="):   firstx  = float(s.split("=",1)[1])
        elif u.startswith("##LASTX="):    lastx   = float(s.split("=",1)[1])
        elif u.startswith("##XYDATA="):
            data_fmt    = s.split("=",1)[1].strip()
            in_xydata   = True
            in_xypoints = False
        elif u.startswith("##XYPOINTS="):
            in_xypoints = True
            in_xydata   = False
        elif u.startswith("##END"):
            in_xydata = in_xypoints = False
        elif in_xydata:
            data_lines.append(s)
        elif in_xypoints:
            xypts_lines.append(s)

    # ----------------------------------------------------------------
    # Format A: XYPOINTS — simple pairs (rare for simulated NMR)
    # ----------------------------------------------------------------
    if xypts_lines:
        xs, ys = [], []
        for line in xypts_lines:
            parts = line.replace(",", " ").split()
            if len(parts) >= 2:
                try:
                    xs.append(float(parts[0]) * xfactor)
                    ys.append(float(parts[1]) * yfactor)
                except ValueError:
                    pass
        return pd.DataFrame({"chemical_shift_ppm": xs, "intensity": ys})

    # ----------------------------------------------------------------
    # Format B: XYDATA=(X++(Y..Y))  — compressed (cheminfo.org default)
    # X on each data line is a RAW INTEGER INDEX into the spectrum array.
    # Gaps between segments are zero (sparse storage of NMR peaks).
    # ----------------------------------------------------------------
    if npoints is None or npoints < 1:
        raise ValueError("NPOINTS not found or invalid")
    if deltax is None:
        if firstx is not None and lastx is not None and npoints > 1:
            deltax = (lastx - firstx) / (npoints - 1)
        else:
            raise ValueError("Cannot determine deltax")

    y_array = np.zeros(npoints, dtype=np.float64)

    for line in data_lines:
        if not line:
            continue
        tokens = tokenize(line)
        if not tokens:
            continue

        # First token = raw X index
        x_idx = int(round(tokens[0][1]))
        if x_idx >= npoints:
            continue  # trailing checksum line (e.g. "40950@")

        pos        = x_idx
        last_y     = 0.0
        last_dif   = 0.0
        in_dif_mode = False

        for t_type, t_val in tokens[1:]:

            if t_type == 'SQZ':
                y = t_val * yfactor
                if pos < npoints:
                    y_array[pos] = y
                pos += 1
                last_y = y
                in_dif_mode = False

            elif t_type == 'DIF':
                diff = t_val * yfactor
                y    = last_y + diff
                if pos < npoints:
                    y_array[pos] = y
                pos += 1
                last_y     = y
                last_dif   = diff
                in_dif_mode = True

            elif t_type == 'DUP':
                count = int(t_val) - 1  # first copy already stored
                if in_dif_mode:
                    for _ in range(count):
                        y = last_y + last_dif
                        if pos < npoints:
                            y_array[pos] = y
                        pos += 1
                        last_y = y
                else:
                    for _ in range(count):
                        if pos < npoints:
                            y_array[pos] = last_y
                        pos += 1

    x_array = np.array([i * deltax for i in range(npoints)], dtype=np.float64)

    return pd.DataFrame({"chemical_shift_ppm": x_array, "intensity": y_array})


def convert_jcamp_to_csv(jdx_path, output_folder):
    stem     = Path(jdx_path).stem
    out_path = os.path.join(output_folder, stem + ".csv")
    df       = parse_jcamp(jdx_path)
    if df.empty or len(df) == 0:
        return False, "No data parsed"
    df.to_csv(out_path, index=False)
    return True, out_path


# ========== MAIN ==========
jdx_files = sorted(glob.glob(os.path.join(SOURCE_FOLDER, "*.jdx")))
total     = len(jdx_files)

log("=" * 60)
log(f"JCAMP-DX → CSV Conversion")
log(f"Found {total} .jdx files in: {SOURCE_FOLDER}")
log(f"Output folder            : {OUTPUT_FOLDER}")
log("=" * 60)

success_count = 0
fail_count    = 0

for i, jdx_path in enumerate(jdx_files, start=1):
    fname = os.path.basename(jdx_path)
    try:
        ok, result = convert_jcamp_to_csv(jdx_path, OUTPUT_FOLDER)
        if ok:
            log(f"[{i:04d}/{total}] OK   | {fname}")
            success_count += 1
        else:
            log(f"[{i:04d}/{total}] FAIL | {fname} — {result}")
            fail_count += 1
    except Exception as e:
        log(f"[{i:04d}/{total}] ERR  | {fname} — {e}")
        fail_count += 1

log("=" * 60)
log(f"Done!  {success_count} converted,  {fail_count} failed.")
log(f"CSVs saved to: {OUTPUT_FOLDER}")
log("=" * 60)
```

```python
import pandas as pd
import os
import glob
import re
from collections import defaultdict

excel_path = r"C:\Users\pljh0187\Desktop\Senushi\Book1.xlsx"
jdx_folder = r"C:\Users\pljh0187\Desktop\Senushi"

df = pd.read_excel(excel_path)
excel_names = df["Name"].astype(str).str.strip().tolist()

def safe(name):
    return "".join(c if c.isalnum() or c in " -_" else "_" for c in name).strip()

excel_safe_map = {safe(n): n for n in excel_names}

# collect frequencies per compound
freqs = defaultdict(set)

files = glob.glob(os.path.join(jdx_folder, "*.jdx"))

for f in files:
    name = os.path.basename(f)
    m = re.match(r"(.+?)_(\d+)MHz\.jdx", name)
    if m:
        base = m.group(1)
        freq = int(m.group(2))
        freqs[base].add(freq)

missing = []

for safe_name, original in excel_safe_map.items():
    fset = freqs.get(safe_name, set())
    if fset != {60, 100, 500}:   # not complete
        missing.append((original, sorted({60, 100, 500} - fset)))

print(f"\nIncomplete compounds: {len(missing)}\n")

for name, missing_freq in missing:
    print(f"{name}  → missing {missing_freq}")
```

```python
import os
import pandas as pd
from datetime import datetime

# ========== CONFIGURATION ==========
CSV_PATH     = r"C:\Users\pljh0187\Desktop\Senushi\Book1.csv"
JDX_FOLDER   = r"C:\Users\pljh0187\Desktop\Senushi"
LOG_FILE     = os.path.join(JDX_FOLDER, "rename_log.txt")
DRY_RUN      = False   # <-- SET TO False TO ACTUALLY RENAME FILES
                      #     Run with True first to preview changes safely
# ====================================

FREQS = [60, 100, 500]

def log(message):
    ts = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    line = f"[{ts}] {message}"
    print(line)
    with open(LOG_FILE, "a", encoding="utf-8") as f:
        f.write(line + "\n")

def old_safe(name):
    """Recreates the mangled filename the automation script produced."""
    safe = "".join(c if c.isalnum() or c in " -_" else "_" for c in name).strip()
    return safe

def safe_new_name(name):
    """Compound name as-is, but replace Windows-illegal colon with semicolon."""
    return name.replace(":", ";")

# ========== LOAD COMPOUNDS ==========
df = pd.read_csv(CSV_PATH)
df["Name"] = df["Name"].str.strip()
compounds = df["Name"].unique().tolist()

log("=" * 70)
log(f"{'DRY RUN — no files will be renamed' if DRY_RUN else 'LIVE RUN — files will be renamed'}")
log(f"Loaded {len(compounds)} unique compound names from: {CSV_PATH}")
log(f"JDX folder : {JDX_FOLDER}")
log("=" * 70)

renamed   = 0
skipped   = 0
not_found = 0
no_change = 0

for name in compounds:
    for freq in FREQS:
        old_name = f"{old_safe(name)}_{freq}MHz.jdx"
        new_name = f"{safe_new_name(name)}_{freq}MHz.jdx"

        old_path = os.path.join(JDX_FOLDER, old_name)
        new_path = os.path.join(JDX_FOLDER, new_name)

        # No change needed (name had no special characters)
        if old_name == new_name:
            no_change += 1
            continue

        # Old file doesn't exist
        if not os.path.exists(old_path):
            log(f"NOT FOUND | {old_name}")
            not_found += 1
            continue

        # Target name already exists (already renamed or duplicate)
        if os.path.exists(new_path):
            log(f"SKIP — target exists | {new_name}")
            skipped += 1
            continue

        # Rename
        log(f"RENAME | '{old_name}'")
        log(f"     -> '{new_name}'")
        if not DRY_RUN:
            os.rename(old_path, new_path)
        renamed += 1

log("=" * 70)
log(f"{'DRY RUN complete' if DRY_RUN else 'Renaming complete'}")
log(f"  Would rename / Renamed : {renamed}")
log(f"  No change needed       : {no_change}")
log(f"  Not found              : {not_found}")
log(f"  Skipped (target exists): {skipped}")
if DRY_RUN:
    log("")
    log("  --> To apply the renames, open the script and set DRY_RUN = False")
log("=" * 70)
```
