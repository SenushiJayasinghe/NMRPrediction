```python
from rdkit import Chem
from rdkit.Chem import AllChem
import pandas as pd

# =========================
# PATH (UPDATED)
# =========================
file_path = r"C:\Users\pljh0187\Desktop\Senushi MestReNova\Book4.csv"
output_sdf = r"C:\Users\pljh0187\Desktop\Senushi MestReNova\compounds.sdf"

# =========================
# LOAD CSV (KEEP 001, 002)
# =========================
df = pd.read_csv(file_path, dtype={"ID": str})

print(f"Loaded {len(df)} compounds")

writer = Chem.SDWriter(output_sdf)

failed = []

# =========================
# LOOP
# =========================
for i, row in df.iterrows():

    smiles = str(row["SMILES"])
    mol_id = str(row["ID"])
    name = str(row["Name"])
    mw = str(row["MW"])

    mol = Chem.MolFromSmiles(smiles)

    if mol is None:
        print(f" Invalid SMILES: {mol_id}")
        failed.append(mol_id)
        continue

    # Build structure
    mol = Chem.AddHs(mol)
    AllChem.EmbedMolecule(mol)
    AllChem.UFFOptimizeMolecule(mol)

    # Store metadata for Mnova
    mol.SetProp("_Name", name)
    mol.SetProp("ID", mol_id)
    mol.SetProp("MW", mw)
    mol.SetProp("SMILES", smiles)

    writer.write(mol)

writer.close()

print("\nDONE")
print(f"SDF saved at: {output_sdf}")
print(f"Failed molecules: {len(failed)}")

if failed:
    print("Failed IDs:", failed)
```
