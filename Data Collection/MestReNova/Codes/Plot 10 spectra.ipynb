```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Read CSV
file = r"C:\Users\pljh0187\Desktop\Senushi MestReNova\Spectra collection\NMR 60 MHz.csv"

df = pd.read_csv(file)

# Chemical shift values are all columns after '#', 'Title', 'Class'
ppm = df.columns[3:].astype(float)

# Randomly choose 10 compounds
random_rows = df.sample(10, random_state=42)

for _, row in random_rows.iterrows():

    intensity = row.iloc[3:].astype(float).values

    plt.figure(figsize=(10,3))

    # Plot as line spectrum
    plt.plot(ppm, intensity, linewidth=1)

    # Reverse x-axis (standard NMR convention)
    plt.gca().invert_xaxis()

    plt.xlabel("Chemical Shift (ppm)")
    plt.ylabel("Intensity")
    plt.title(f"Compound {row['#']}")

    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load CSV file
file = r"C:\Users\pljh0187\Desktop\Senushi MestReNova\Spectra collection\NMR 90 MHz.csv"
df = pd.read_csv(file)

# Extract ppm axis (assumes first 3 columns are metadata)
ppm = df.columns[3:].astype(float)

# Randomly select 10 compounds (true random each run)
random_rows = df.sample(10)

# Plot each selected spectrum
for _, row in random_rows.iterrows():

    # Extract raw intensities (NO normalization)
    intensity = row.iloc[3:].astype(float).values

    plt.figure(figsize=(10, 3))

    # Plot spectrum
    plt.plot(ppm, intensity, linewidth=1)

    # Standard NMR convention
    plt.gca().invert_xaxis()

    # Labels and title
    plt.xlabel("Chemical Shift (ppm)")
    plt.ylabel("Intensity (raw)")
    plt.title(f"Compound {row['#']}")

    plt.tight_layout()
    plt.show()
```
